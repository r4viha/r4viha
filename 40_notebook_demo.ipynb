```python
from random import randint
from statistics import mean

numbers = []

for _ in range(10000):
    numbers.append(randint(0, 100))

print(mean(numbers))
```

```python
def has_punctuation(passwd):
    from string import punctuation

    for character in passwd:
        if character in punctuation:
            return True

    return False

def has_digit(passwd):
    from string import digits

    for character in passwd:
        if character in digits:
            return True

    return False

for test_string in ['banana!', 'banana', 'x.', '123456#', 'cheesey123']:
    print(f'{test_string} -> {has_punctuation(test_string)}')
    print(f'{test_string} -> {has_digit(test_string)}')
```

```python
def has_punctuation(passwd):
    from string import punctuation

    for character in passwd:
        if character in punctuation:
            return True

    return False

def has_digit(passwd):
    from string import digits

    for character in passwd:
        if character in digits:
            return True

    return False

def is_valid_password(passwd):
    return has_punctuation(passwd) and has_digit(passwd)

for test_string in ['banana!', 'banana', 'x.', '123456#', 'cheesey123', '911', '#911']:
    print(f'{test_string} -> {is_valid_password(test_string)}')
```
